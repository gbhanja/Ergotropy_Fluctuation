```python
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import os


downloads = os.path.join(os.path.expanduser("~"), "Downloads")

filename = os.path.join(downloads, "CS_Kac_OP_bosonic.npz")

data = np.load(filename)

# extract arrays
N_arr2 = data["N"]
tau2 = data["tau"]
E_ergo2 = data["Eerg"]
E_var2 = data["variance"]
E_ratio2 = data["ratio"]
E_B2 = data["Eb"]
Gnd_ex1 = data["gnd_excitation"]
Res_ex1 = data["res_excitation"]

# derived quantities
Power2 = E_B2 / tau2


F_ratio2 = np.sqrt(E_var2) / E_ergo2
FE_ratio2 = np.sqrt(E_var2) / E_B2
```

```python
##########################################################
# B(N), a(N), b(N) — vectorized over all N
##########################################################

ω0 = 1.0

# Second moments (vectorized over the whole array)
E2_sym  = E_var2 + E_ergo2**2     # <E_sym^2>   (indistinguishable / bosonic)
E2_full = E_var1 + E_ergo1**2     # <E_full^2>  (distinguishable / full)

# Exact discriminant B(N) = <E_full>^2<E_sym^2> - <E_sym>^2<E_full^2>
B_arr = E_ergo1**2 * E2_sym - E_ergo2**2 * E2_full

for i in range(len(N_arr1)):
    print(
        f"N={N_arr1[i]}, "
        f"B={B_arr[i]:.6e}"
    )

##########################################################
# Extract dimensionless coefficients a and b (vectorized)
#   a = (<E_full^2> - <E_sym^2>) / ω0^2
#   b = (<E_full> - <E_sym>) / ω0
##########################################################

b_arr = (E_ergo1 - E_ergo2) / ω0
a_arr = (E2_full - E2_sym) / ω0**2

for i in range(len(N_arr1)):
    print(
        f"N={N_arr1[i]}, "
        f"a={a_arr[i]:.6f}, "
        f"b={b_arr[i]:.6f}, "
        f"a/b={a_arr[i]/b_arr[i] if b_arr[i]!=0 else float('nan'):.6f}"
    )

##########################################################
# Exact-condition components:
#   RHS = ω0 * a/b               (cost)
#   LHS = 2<E_sym> + ω0 b        (resource, sufficient form)
##########################################################

safe_b = np.where(np.abs(b_arr) < 1e-12, np.nan, b_arr)

RHS_cost     = ω0 * a_arr / safe_b
LHS_resource = 2 * E_ergo2 + ω0 * b_arr

g_sym    = E2_sym / E_ergo2**2          # relative 2nd-moment factor (>=1)
LHS_exact = g_sym * LHS_resource        # exact LHS with fluctuation factor

##########################################################
# Plot 1: ω a/b  vs  2<E_sym> + ω b  (scatter, colored by N)
##########################################################

plt.figure(figsize=(14, 6), dpi=500)

from matplotlib.colors import Normalize

norm = Normalize(vmin=N_arr1.min(), vmax=N_arr1.max())

sc = plt.scatter(
    LHS_resource,
    RHS_cost,
    c=N_arr1,
    cmap='viridis',
    norm=norm,
    marker='o',
    s=60,
    edgecolors='k',
    linewidth=0.4,
    label='Data points (per N)'
)

# Equality (threshold) line
lims = [0, np.nanmax([np.nanmax(LHS_resource), np.nanmax(RHS_cost)]) * 1.05]
plt.plot(lims, lims, 'r--', lw=2, label='Equality line (threshold)')

plt.xlabel(r'$2\langle\mathcal{E}_{\rm sym}\rangle + \omega_0 b$  (resource)', fontsize=13)
plt.ylabel(r'$\omega_0\, a/b$  (cost)', fontsize=13)
plt.title(
    r'Exact condition test: points below the line satisfy '
    r'$\sqrt{\Delta\mathcal{E}_{\rm sym}^2}/\langle\mathcal{E}_{\rm sym}\rangle '
    r'\geq \sqrt{\Delta\mathcal{E}_{\rm full}^2}/\langle\mathcal{E}_{\rm full}\rangle$',
    fontsize=11
)

cbar = plt.colorbar(sc, pad=0.02)
cbar.set_label(r'$N$', fontsize=12)

plt.legend(fontsize=10)
plt.tight_layout()
plt.show()

##########################################################
# Plot 2: Resource vs Cost vs N (direct comparison)
##########################################################

plt.figure(figsize=(14, 6), dpi=500)

plt.plot(N_arr1, RHS_cost, 'r^-', lw=2, ms=7, label=r'$\omega_0\, a/b$ (cost)')
plt.plot(N_arr1, LHS_resource, 'bo-', lw=2, ms=7, label=r'$2\langle\mathcal{E}_{\rm sym}\rangle+\omega_0 b$ (resource)')
plt.plot(N_arr1, LHS_exact, 'g.--', lw=2, ms=7, label=r'$g_{\rm sym}(2\langle\mathcal{E}_{\rm sym}\rangle+\omega_0 b)$ (exact)')

plt.xlabel(r'$N$', fontsize=13)
plt.ylabel(r'Energy (units of $\omega_0$)', fontsize=13)
plt.title(
    rf'Exact condition components vs $N$ for the {model} model, $\omega_0={ω0}$',
    fontsize=12
)
plt.legend()
plt.tight_layout()
plt.show()

##########################################################
# Plot 3: Log-log scaling comparison
##########################################################

# Note's sufficient-condition bound: ω0 * m_tilde = ω0*(N(N+1)/2 - m_0)
# Using ground-state excitation (Gnd_ex1) as m_0
note_bound = ω0 * ((N_arr1 * (N_arr1 + 1)) / 2.0 - Gnd_ex1)

valid = (LHS_resource > 0) & (np.abs(RHS_cost) > 0) & (note_bound > 0)

logN            = np.log(N_arr1[valid])
log_resource    = np.log(LHS_resource[valid])
log_cost        = np.log(np.abs(RHS_cost[valid]))
log_note_bound  = np.log(note_bound[valid])

slope_resource, intercept_resource = np.polyfit(logN, log_resource, 1)
slope_cost, intercept_cost = np.polyfit(logN, log_cost, 1)
slope_note, intercept_note = np.polyfit(logN, log_note_bound, 1)

print(f"\nScaling fit: resource   ~ N^{slope_resource:.3f}")
print(f"Scaling fit: cost       ~ N^{slope_cost:.3f}")
print(f"Scaling fit: note bound ~ N^{slope_note:.3f}")
print("(If slope_cost/slope_resource are close to 1 while slope_note is close to 2,")
print(" this confirms the note's O(N^2) bound can be relaxed to O(N).)")

fig, ax = plt.subplots(figsize=(9, 7), dpi=500)

ax.plot(logN, log_resource, 'bo-', ms=7, lw=2, label=r'$2\langle\mathcal{E}_{\rm sym}\rangle+\omega_0 b$')
ax.plot(logN, log_cost, 'r^-', ms=7, lw=2, label=r'$\omega_0\, a/b$')
ax.plot(logN, log_note_bound, 'k:', lw=2, label=r'bound $\omega_0\tilde m$ (reference)')

ax.set_xlabel(r'$\log(N)$')
ax.set_ylabel(r'$\log(\mathrm{Energy\ (units\ of\ \omega_0)})$')
ax.set_title(
    rf'Log-log scaling: resource vs cost vs original $O(N^2)$ bound '
    rf'({model} model, $\omega_0={ω0}$)',
    fontsize=13
)

logN_ticks = np.log(N_arr1[valid][::len(N_arr1[valid])//8])  # sparse ticks
ax.set_xticks(logN_ticks)
ax.set_xticklabels([f'{v:.1f}' for v in logN_ticks], fontsize=11)
ax.tick_params(axis='y', labelsize=11)

ax.legend(fontsize=10, frameon=True, loc='best')
plt.tight_layout()
plt.show()

##########################################################
# Print summary table
##########################################################

print(f"\n{'N':>4} {'a':>12} {'b':>12} {'ωa/b':>14} {'2E_{sym}+ωb':>14}"
      f"{'exact LHS':>14} {'note bound':>14} {'Holds?':>8}")
for i in range(len(N_arr1)):
    holds = LHS_exact[i] >= RHS_cost[i] if not np.isnan(RHS_cost[i]) else True
    print(f"{N_arr1[i]:4d} {a_arr[i]:12.4f} {b_arr[i]:12.4f} {RHS_cost[i]:14.4f} "
          f"{LHS_resource[i]:14.4f} {LHS_exact[i]:14.4f} {note_bound[i]:14.4f} "
          f"{str(holds):>8}")
```

```python
##############################################################
# log(Relative ergotropy fluctuations) vs log(N)
# for omega_0 = 1.0
##############################################################

import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# Fixed omega_0
# ============================================================

omega0 = 1.0


# ============================================================
# Selected N values
# ============================================================

N_selected = np.array([
    2, 10, 50, 100
])


# ============================================================
# Maximum battery energy
# ============================================================

E_max1 = omega0 * N_arr1
E_max2 = omega0 * N_arr2


# ============================================================
# Normalized ergotropy
# ============================================================

x1 = E_ergo1 / E_max1
x2 = E_ergo2 / E_max2


# ============================================================
# Select requested N values
# ============================================================

mask1 = np.isin(
    N_arr1,
    N_selected
)

mask2 = np.isin(
    N_arr2,
    N_selected
)

N1 = N_arr1[mask1]
F1 = F_ratio1[mask1]
x1_sel = x1[mask1]

N2 = N_arr2[mask2]
F2 = F_ratio2[mask2]
x2_sel = x2[mask2]


# ============================================================
# Sort according to N
# ============================================================

i1 = np.argsort(N1)

N1 = N1[i1]
F1 = F1[i1]
x1_sel = x1_sel[i1]

i2 = np.argsort(N2)

N2 = N2[i2]
F2 = F2[i2]
x2_sel = x2_sel[i2]


# ============================================================
# Keep only positive values for logarithm
# ============================================================

valid1 = (
    (N1 > 0) &
    (F1 > 0)
)

valid2 = (
    (N2 > 0) &
    (F2 > 0)
)


# ============================================================
# Explicit logarithmic transformation
# ============================================================

logN1 = np.log(N1[valid1])
logF1 = np.log(F1[valid1])

logN2 = np.log(N2[valid2])
logF2 = np.log(F2[valid2])


# ============================================================
# Theoretical bounds
# ============================================================

valid_bound = (
    (x1_sel > 0) &
    (x1_sel <= 1)
)

N_bound = N1[valid_bound]
x_bound = x1_sel[valid_bound]


general_bound = np.sqrt(
    1.0 / x_bound**2 - 1.0
)

positive_work_bound = np.sqrt(
    1.0 / x_bound - 1.0
)


# ============================================================
# Keep only positive bounds for logarithm
# ============================================================

valid_general = general_bound > 0
valid_positive = positive_work_bound > 0


# ============================================================
# Figure
# ============================================================

fig, ax = plt.subplots(
    figsize=(10, 7),
    dpi=300
)


# ============================================================
# Distinguishable battery
# ============================================================

ax.plot(
    logN1,
    logF1,
    'o-',
    ms=7,
    lw=2.0,
    label='Distinguishable battery',
    zorder=4
)


# ============================================================
# Indistinguishable battery
# ============================================================

ax.plot(
    logN2,
    logF2,
    's--',
    ms=7,
    lw=2.0,
    label='Indistinguishable battery',
    zorder=4
)


# ============================================================
# General upper bound
# ============================================================

ax.plot(
    np.log(N_bound[valid_general]),
    np.log(general_bound[valid_general]),
    color='black',
    linestyle='-',
    linewidth=2.0,
    label='General upper bound',
    zorder=2
)


# ============================================================
# Positive-work upper bound
# ============================================================

ax.plot(
    np.log(N_bound[valid_positive]),
    np.log(positive_work_bound[valid_positive]),
    color='black',
    linestyle='--',
    linewidth=2.0,
    label='Positive-work upper bound',
    zorder=2
)


# ============================================================
# Axis labels
# ============================================================

ax.set_xlabel(
    r'$\log(N)$',
    fontsize=15
)

ax.set_ylabel(
    r'$\log\!\left('
    r'\sigma_{\mathcal{E}_b}/\langle\mathcal{E}_b\rangle'
    r'\right)$',
    fontsize=15
)


# ============================================================
# Title
# ============================================================

ax.set_title(
    rf'Log-relative ergotropy fluctuations vs $\log(N)$ '
    rf'of the {model} model for $\omega_0={omega0}$',
    fontsize=14
)


# ============================================================
# X ticks 
# ============================================================

logN_ticks = np.log(N_selected)

ax.set_xticks(logN_ticks)

ax.set_xticklabels(
    [f'{np.log(N):.1f}' for N in N_selected],
    fontsize=11
)


# ============================================================
# Tick parameters
# ============================================================

ax.tick_params(
    axis='y',
    labelsize=11
)



# ============================================================
# Legend
# ============================================================

ax.legend(
    fontsize=10,
    frameon=True,
    loc='best'
)


# ============================================================
# Layout
# ============================================================

plt.tight_layout()

plt.show()
```

```python
##############################################################
# Difference in log(relative ergotropy fluctuations)
# between distinguishable and indistinguishable batteries
# as a function of log(N)
# for omega_0 = 1.0
##############################################################

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D


# ============================================================
# Fixed omega_0
# ============================================================

omega0 = 1.0


# ============================================================
# Selected N values
# ============================================================

N_selected = np.array([
    2, 10, 50, 100
])


# ============================================================
# Select requested N values
# ============================================================

mask1 = np.isin(
    N_arr1,
    N_selected
)

mask2 = np.isin(
    N_arr2,
    N_selected
)

N1 = N_arr1[mask1]
F1 = F_ratio1[mask1]

N2 = N_arr2[mask2]
F2 = F_ratio2[mask2]


# ============================================================
# Sort according to N
# ============================================================

i1 = np.argsort(N1)

N1 = N1[i1]
F1 = F1[i1]


i2 = np.argsort(N2)

N2 = N2[i2]
F2 = F2[i2]


# ============================================================
# Find common N values
# ============================================================

common_N = np.intersect1d(
    N1,
    N2
)


# ============================================================
# Extract F values at common N
# ============================================================

F_dist = np.array([
    F1[N1 == N][0]
    for N in common_N
])

F_indist = np.array([
    F2[N2 == N][0]
    for N in common_N
])


# ============================================================
# Keep only positive fluctuations
# ============================================================

valid = (
    (common_N > 0) &
    (F_dist > 0) &
    (F_indist > 0)
)

N_common = common_N[valid]

F_dist = F_dist[valid]
F_indist = F_indist[valid]


# ============================================================
# Logarithms
# ============================================================

logN = np.log(N_common)

logF_dist = np.log(F_dist)

logF_indist = np.log(F_indist)


# ============================================================
# Difference in logarithmic fluctuations
# ============================================================

delta_logF = (
    logF_dist - logF_indist
)


# ============================================================
# Figure
# ============================================================

fig, ax = plt.subplots(
    figsize=(9, 7),
    dpi=500
)


# ============================================================
# Main curve
# ============================================================

ax.plot(
    logN,
    delta_logF,
    'o-',
    color='black',
    markersize=8,
    linewidth=2,
    label=r'$\log(\mathcal{F}_{\rm dist})-\log(\mathcal{F}_{\rm indist})$'
)


# ============================================================
# Zero reference
# ============================================================

ax.axhline(
    0,
    color='red',
    linestyle='--',
    linewidth=1.5,
    label=r'$\Delta\log \mathcal{F}=0$'
)


# ============================================================
# Axis labels
# ============================================================

ax.set_xlabel(
    r'$\log(N)$',
    fontsize=15
)

ax.set_ylabel(
    r'$\log(\mathcal{F}_{\rm dist})-\log(\mathcal{F}_{\rm indist}),$'
    r'$\mathcal{F} = \sigma_{\mathcal{E}_b}/\langle\mathcal{E}_b\rangle $',
    fontsize=15
)


# ============================================================
# Title
# ============================================================

ax.set_title(
    rf'Difference in logarithmic relative ergotropy '
    rf'fluctuations of the {model} model for $\omega_0={omega0}$',
    fontsize=14
)


# ============================================================
# X-axis ticks
# ============================================================

logN_ticks = np.log(N_selected)

ax.set_xticks(logN_ticks)

ax.set_xticklabels(
    [f'{np.log(N):.1f}' for N in N_selected],
    fontsize=11
)

# ============================================================
# Tick parameters
# ============================================================

ax.tick_params(
    axis='y',
    labelsize=11
)


# ============================================================
# Legend
# ============================================================

ax.legend(
    fontsize=10,
    frameon=True,
    loc='best'
)


# ============================================================
# Layout
# ============================================================

plt.tight_layout()

plt.show()
```

```python
##############################################################
# Vertical subplots for omega_0 = 1.0
#
# Top:
# log(relative ergotropy fluctuation) vs log(N)
#
# Bottom:
# difference between distinguishable and indistinguishable
# logarithmic relative fluctuations vs log(N)
##############################################################

import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# Fixed omega_0
# ============================================================

omega0 = 1.0


# ============================================================
# Selected N values
# ============================================================

N_selected = np.array([
    2, 10, 50, 100
])


# ============================================================
# Maximum battery energy
# ============================================================

E_max1 = omega0 * N_arr1
E_max2 = omega0 * N_arr2


# ============================================================
# Normalized ergotropy
# ============================================================

x1 = E_ergo1 / E_max1
x2 = E_ergo2 / E_max2


# ============================================================
# Select requested N values
# ============================================================

mask1 = np.isin(N_arr1, N_selected)
mask2 = np.isin(N_arr2, N_selected)

N1 = N_arr1[mask1]
F1 = F_ratio1[mask1]
x1_sel = x1[mask1]

N2 = N_arr2[mask2]
F2 = F_ratio2[mask2]
x2_sel = x2[mask2]


# ============================================================
# Sort according to N
# ============================================================

i1 = np.argsort(N1)

N1 = N1[i1]
F1 = F1[i1]
x1_sel = x1_sel[i1]

i2 = np.argsort(N2)

N2 = N2[i2]
F2 = F2[i2]
x2_sel = x2_sel[i2]


# ============================================================
# Keep only positive values for logarithms
# ============================================================

valid1 = (
    (N1 > 0) &
    (F1 > 0)
)

valid2 = (
    (N2 > 0) &
    (F2 > 0)
)


# ============================================================
# Logarithmic quantities
# ============================================================

logN1 = np.log(N1[valid1])
logF1 = np.log(F1[valid1])

logN2 = np.log(N2[valid2])
logF2 = np.log(F2[valid2])


# ============================================================
# Common N values for difference plot
# ============================================================

common_N = np.intersect1d(N1, N2)

F_dist = np.array([
    F1[N1 == N][0]
    for N in common_N
])

F_indist = np.array([
    F2[N2 == N][0]
    for N in common_N
])


valid_diff = (
    (common_N > 0) &
    (F_dist > 0) &
    (F_indist > 0)
)

N_common = common_N[valid_diff]

F_dist = F_dist[valid_diff]
F_indist = F_indist[valid_diff]


# ============================================================
# Logarithmic difference
# ============================================================

logN_diff = np.log(N_common)

logF_dist = np.log(F_dist)
logF_indist = np.log(F_indist)

delta_logF = logF_dist - logF_indist


# ============================================================
# Theoretical bounds
# ============================================================

valid_bound = (
    (x1_sel > 0) &
    (x1_sel <= 1)
)

N_bound = N1[valid_bound]
x_bound = x1_sel[valid_bound]


general_bound = np.sqrt(
    1.0 / x_bound**2 - 1.0
)

positive_work_bound = np.sqrt(
    1.0 / x_bound - 1.0
)


valid_general = general_bound > 0
valid_positive = positive_work_bound > 0


# ============================================================
# Create vertical subplots
# ============================================================

fig, axes = plt.subplots(
    2,
    1,
    figsize=(10, 12),
    dpi=500,
    sharex=True,
    gridspec_kw={'height_ratios': [1, 1]}
)

ax1, ax2 = axes


# ============================================================
# TOP PLOT
# log(relative fluctuation) vs log(N)
# ============================================================

ax1.plot(
    logN1,
    logF1,
    'o-',
    ms=7,
    lw=2.0,
    label='Distinguishable battery',
    zorder=4
)

ax1.plot(
    logN2,
    logF2,
    's--',
    ms=7,
    lw=2.0,
    label='Indistinguishable battery',
    zorder=4
)


# ------------------------------------------------------------
# General upper bound
# ------------------------------------------------------------

ax1.plot(
    np.log(N_bound[valid_general]),
    np.log(general_bound[valid_general]),
    color='black',
    linestyle='-',
    linewidth=2.0,
    label='General upper bound',
    zorder=2
)


# ------------------------------------------------------------
# Positive-work upper bound
# ------------------------------------------------------------

ax1.plot(
    np.log(N_bound[valid_positive]),
    np.log(positive_work_bound[valid_positive]),
    color='black',
    linestyle='--',
    linewidth=2.0,
    label='Positive-work upper bound',
    zorder=2
)


# ------------------------------------------------------------
# Top plot title
# ------------------------------------------------------------

ax1.set_title(
    rf'Log-relative ergotropy fluctuations vs $\log(N)$ ',
    fontsize=14
)


# ------------------------------------------------------------
# Top y-axis
# ------------------------------------------------------------

ax1.set_ylabel(
    r'$ \mathcal{F} = \sigma_{\mathcal{E}_b}/\langle\mathcal{E}_b\rangle $',
    fontsize=15
)


# ------------------------------------------------------------
# Top legend
# ------------------------------------------------------------

ax1.legend(
    fontsize=9,
    frameon=True,
    loc='best'
)


# ============================================================
# BOTTOM PLOT
# Difference between the logarithmic fluctuations
# ============================================================

ax2.plot(
    logN_diff,
    delta_logF,
    'o-',
    color='black',
    ms=7,
    lw=2.0,
    label=r'$\log(\mathcal{F}_{\rm dist})-\log(\mathcal{F}_{\rm indist})$'
)


# ------------------------------------------------------------
# Zero reference
# ------------------------------------------------------------

ax2.axhline(
    0,
    color='red',
    linestyle='--',
    linewidth=1.5,
    label=r'$\Delta\log(\mathcal{F})=0$'
)

# ------------------------------------------------------------
# Bottom plot title
# ------------------------------------------------------------

ax2.set_title(
    r'Difference in logarithmic fluctuations',
    fontsize=14
)

# ------------------------------------------------------------
# Bottom y-axis
# ------------------------------------------------------------

ax2.set_ylabel(
    r'$\log(\mathcal{F}_{\rm dist})-\log(\mathcal{F}_{\rm indist})$',
    fontsize=15
)


# ============================================================
# Common x-axis
# ============================================================

ax2.set_xlabel(
    r'$\log(N)$',
    fontsize=15
)


# ============================================================
# X-axis ticks
# ============================================================

logN_ticks = np.log(N_selected)

for ax in axes:

    ax.set_xticks(logN_ticks)

    ax.set_xticklabels(
        [f'{np.log(N):.1f}' for N in N_selected],
        fontsize=11
    )

    ax.tick_params(
        axis='y',
        labelsize=11
    )


# ============================================================
# Bottom legend
# ============================================================

ax2.legend(
    fontsize=9,
    frameon=True,
    loc='best'
)


# ============================================================
# Overall title
# ============================================================

fig.suptitle(
    rf'Relative ergotropy fluctuations in the {model} model for $\omega_0={omega0}$',
    fontsize=17,
    y=0.98
)


# ============================================================
# Layout
# ============================================================

plt.tight_layout(
    rect=[0.05, 0.04, 0.98, 0.95]
)

plt.show()
```
